#### Join with `gdf_stops` — why SNCF stop count differs

`accessibilite-en-gare.csv` uses `stop_point:IDFM:monomodalStopPlace:XXXXX` as identifier.  
Strip `stop_point:` to match `stop_id` in `arrets-lignes.geojson`.

```python
print(f"Number of SNCF stops in gdf_stops = {gdf_stops.query("operatorname=='SNCF'")['stop_id'].nunique()}")
print(f"Number of stops in df_access_2 = {df_access_2['stop_point_id'].nunique()}")
```

```python
sncf_stops = gdf_stops[gdf_stops["operatorname"] == "SNCF"].copy()

df_access_2_norm = df_access_2.copy()
df_access_2_norm["stop_id"] = df_access_2_norm["stop_point_id"].str.replace("stop_point:", "", regex=False)

merged = sncf_stops.merge(
    df_access_2_norm[["stop_id", "accessibility_level_name"]],
    on="stop_id",
    how="left",
    indicator=True,
)
merged["matched"] = merged["_merge"] == "both"

# Mode distribution: matched vs unmatched explains the count discrepancy
merged.groupby(["mode", "matched"]).size().unstack(fill_value=0).rename(
    columns={True: "matched", False: "unmatched"}
)
```
