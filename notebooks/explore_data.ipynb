```python
geoplotter = GeoPlotter(gdf, geom_col="geometry")
geoplotter.add_geodata_to_map()
geoplotter.map
```

```python
gdf_tmp = gdf.query("`route_long_name` == '14'")

geoplotter_2 = GeoPlotter(gdf_tmp, geom_col="geometry", tiles="Cartodb Positron")
geoplotter_2.add_geodata_to_map()
geoplotter_2.map
```

```python
route_type_colors = {
    "Bus":       "#888888",
    "CableWay":  "#9B59B6",
    "Funicular": "#E67E22",
    "Rail":      "#2471A3",
    "Subway":    "#E74C3C",
    "Tram":      "#27AE60",
}

geoplotter_all = GeoPlotter(gdf, geom_col="geometry", tiles="Cartodb Positron")
geoplotter_all.add_geodata_to_map(
    color_col="route_type",
    colormap=route_type_colors,
    tooltip_cols=["route_short_name", "route_long_name", "route_type", "operatorname"],
)
geoplotter_all.map
```
